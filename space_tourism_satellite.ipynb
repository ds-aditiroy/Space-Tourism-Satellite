# VISUALISATION

```python
plt.figure(figsize=(5,5)) #optional
plt.hist(spc['Fuel_Efficiency_Score'],color='yellow',edgecolor='blue')
plt.title("Histogram")
plt.xlabel("Fuel_Efficiency_Score")
plt.ylabel("frequnecy")
plt.show()
```

### Highlights the distribution of operational efficiency, revealing whether top-performing scores are common or heavily skewed toward lower efficiency tiers.

```python
sns.kdeplot(spc['Cost_Per_Kg'],fill=True,color='purple')
plt.title(("KDE Plot"),color='green')
plt.xlabel("Cost_Per_Kg")
plt.ylabel("frequnecy")
plt.show()
```

### Right-Skewed Distribution: The distribution of Cost_Per_Kg is heavily right-skewed (positively skewed), with a sharp peak concentrated at lower values and a long tail extending toward higher costs.
### Cost Concentration: Most launch missions have a relatively low cost per kilogram, while only a few extreme instances feature significantly high cost per kilogram values (outliers/extreme cases).

```python
plt.figure()
sns.boxplot(spc['Launch_Cost_USD_M'],orient="v",color='red')
plt.title("Box Plot")
plt.xlabel("Efficiency_Category")
plt.show()
```

### Evaluates cost differences between High & Normal efficiency tiers,testing wheather superior fuel efficiency correlates with premium pricing or operational savings.

```python
plt.figure()
spc['Orbit_Type'].value_counts().plot.pie(autopct='%1.1f%%')
plt.title("pie chart")
plt.show()
```

### Shows the market share of missions destined for each orbital path,revealing which region dominates commercial traffic volume.

```python
plt.figure()
sns.barplot(x="Orbit_Type",y="Cost_Per_Kg",data=spc,color='green')
plt.title(("bar plot"),color='green')
plt.show()
```

### Categorizes launches by Mission Type such as Low Earth Orbit(LEO),Sub-orbital , and Medium Earth Orbit(MEO) to identify which sector spent how much cost per kg of fuel.

```python
sns.scatterplot(
    x="Payload_Weight_kg", 
    y="Launch_Cost_USD_M", 
    hue="Orbit_Type", 
    data=spc, 
    palette="Set1"
)
plt.legend(bbox_to_anchor=(1.05,1),loc='upper left')
plt.title("Payload Weight vs Launch Cost by Orbit Type")
plt.show()
```

### No Linear Relationship: Payload_Weight_kg and Launch_Cost_USD_M show zero correlation, with data points uniformly scattered across the plot.
### Even Categorical Distribution: All categories of Orbit_Type (LEO, MEO, GEO, Sub-orbital, LTO) are evenly mixed across all price ranges and payload weights.
### Synthetic Data Behavior: Orbit type does not create distinct clusters or cost tiers, indicating randomly generated data with no underlying real-world patterns.

```python
corr = spc.corr(numeric_only=True)
corr.round(2)
```

```python
sns.heatmap(corr,annot=True)
plt.show()
```

### Strong Negative Correlation: **Payload_Weight_kg** and **Cost_Per_Kg** have a moderate-to-strong negative correlation (-0.57), meaning as payload weight increases, cost per kilogram decreases due to economies of scale.
 ### Weak Positive Correlation: **Launch_Cost_USD_M** and **Cost_Per_Kg** show a moderate positive correlation (0.32), indicating that higher overall launch costs correspond to higher per-kg costs.
### No Linear Relationship Among Most Features: Most other feature pairs show near-zero correlation values (ranging around 0.00), confirming that the majority of variables in this dataset operate independently of one another.

### Project Conclusion: Space Tourism & Satellite Data Analysis
### Data Characteristics: The analysis indicates that most continuous variables (such as **payload weight**, **fuel efficiency**, and **launch cost**) exhibit random, uniform distributions with near-zero mutual correlation, typical of a synthetic dataset.
### Cost Efficiency Insight: A notable negative correlation (-0.57) exists between **Payload Weight** and **Cost Per Kg**, demonstrating clear economies of scale where heavier launch payloads significantly reduce per-unit shipping costs.
### Cost Distribution: The distribution of **Cost Per Kg** is heavily right-skewed, showing that while most missions operate within a lower, cost-efficient range, a small subset of specialized or low-weight launches incur disproportionately higher per-kilogram expenses.
### Strategic Takeaway: Optimization in space launch operations relies heavily on maximizing payload capacity per launch to achieve optimal cost-to-weight efficiency across all orbit types.
